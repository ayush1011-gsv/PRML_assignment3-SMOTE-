1. Download the dataset from https://www.kaggle.com/datasets/uciml/iris. Create a linear
regression model to predict the categories. Create training, validation, and test splits in
the dataset taking 80:10:10 and 70:15:15 proportions.


```python
import pandas as pd
import numpy as np

df=pd.read_csv("/content/Iris.csv")
X = df.drop(["Id", "Species"], axis=1).values

y = df["Species"].map({
    "Iris-setosa": 0,
    "Iris-versicolor": 1,
    "Iris-virginica": 2
}).values

print("\nFeatures shape:", X.shape)
print("Target shape:", y.shape)

def split_data(X, y, train_ratio, val_ratio, test_ratio):
    # Shuffle dataset indices
    indices = np.random.permutation(len(X))
    X = X[indices] #x= input feature
    y = y[indices] #y=target categories
    # Calculate split sizes
    train_size = int(train_ratio * len(X))
    val_size = int(val_ratio * len(X))
    # Split dataset
    X_train = X[0:train_size:1]
    y_train = y[0:train_size:1]
    X_val = X[train_size : train_size + val_size :1]
    y_val = y[train_size : train_size + val_size:1]

    X_test = X[train_size + val_size:]
    y_test = y[train_size + val_size:]

    return X_train, X_val, X_test, y_train, y_val, y_test

class LinearRegression:

    def fit(self, X, y):

        # Add a column of ones for intercept
        X = np.c_[np.ones(X.shape[0]), X]

        # Normal equation
        self.weights = np.linalg.pinv(X.T @ X) @ X.T @ y

    def predict(self, X):

        # Add intercept column
        X = np.c_[np.ones(X.shape[0]), X]

        # Calculate predictions
        return X @ self.weights

# ==========================================
# 6. CREATE OUR OWN EVALUATION FUNCTIONS
# ==========================================

def accuracy(y_actual, y_pred):

    correct = np.sum(y_actual == y_pred)

    return correct / len(y_actual)


def mse(y_actual, y_pred):

    return np.mean((y_actual - y_pred) ** 2)


# ==========================================
# 7. TRAIN AND EVALUATE MODEL
# ==========================================

def train_model(X, y, train_ratio, val_ratio, test_ratio):

    X_train, X_val, X_test, y_train, y_val, y_test = split_data(
        X, y, train_ratio, val_ratio, test_ratio
    )

    print("\n================================")
    print(f"Split: {train_ratio}:{val_ratio}:{test_ratio}")
    print("================================")

    print("Training samples:", len(X_train))
    print("Validation samples:", len(X_val))
    print("Testing samples:", len(X_test))

    # Create model
    model = LinearRegression()

    # Train model
    model.fit(X_train, y_train)

    # Make predictions
    train_pred = model.predict(X_train)
    val_pred = model.predict(X_val)
    test_pred = model.predict(X_test)

    # Convert continuous predictions into categories
    train_pred = np.clip(np.rint(train_pred), 0, 2).astype(int)
    val_pred = np.clip(np.rint(val_pred), 0, 2).astype(int)
    test_pred = np.clip(np.rint(test_pred), 0, 2).astype(int)

    # Calculate accuracy
    train_acc = accuracy(y_train, train_pred)
    val_acc = accuracy(y_val, val_pred)
    test_acc = accuracy(y_test, test_pred)

    # Calculate MSE
    error = mse(y_test, test_pred)

    print("\nTraining Accuracy:", train_acc)
    print("Validation Accuracy:", val_acc)
    print("Testing Accuracy:", test_acc)

    print("Testing MSE:", error)

    return {
        "Split": f"{train_ratio}:{val_ratio}:{test_ratio}",
        "Training Accuracy": train_acc,
        "Validation Accuracy": val_acc,
        "Testing Accuracy": test_acc,
        "MSE": error
    }


# ==========================================
# 8. FIRST SPLIT: 80:10:10
# ==========================================

result1 = train_model(X, y, 0.80, 0.10, 0.10)


# ==========================================
# 9. SECOND SPLIT: 70:15:15
# ==========================================

result2 = train_model(X, y, 0.70, 0.15, 0.15)


# ==========================================
# 10. COMPARE RESULTS
# ==========================================

comparison = pd.DataFrame([result1, result2])

print("\nFINAL COMPARISON")
print(comparison)
```

2. Use oversampling techniques to balance the dataset for each category. Create a linear
regression model on this oversampled dataset. Do the experiment for each of the
categories and create classifiers. Compare their performance. Use stochastic gradient descent.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# ==========================================
# 1. LOAD DATASET
# ==========================================

df = pd.read_csv("Iris.csv")
X = df.drop(["Id", "Species"], axis=1).values
y = df["Species"].map({
    "Iris-setosa": 0,
    "Iris-versicolor": 1,
    "Iris-virginica": 2
}).values
# ==========================================
# 2. SPLIT DATASET
# ==========================================
def split_data(X, y):
    train_indices = []
    val_indices = []
    test_indices = []
    # Stratified splitting
    for category in np.unique(y):

        indices = np.where(y == category)[0]

        np.random.shuffle(indices)

        n = len(indices)

        train_size = int(0.80 * n)
        val_size = int(0.10 * n)

        train_indices.extend(indices[:train_size])
        val_indices.extend(indices[train_size:train_size + val_size])
        test_indices.extend(indices[train_size + val_size:])

    np.random.shuffle(train_indices)
    np.random.shuffle(val_indices)
    np.random.shuffle(test_indices)

    return (
        X[train_indices],
        X[val_indices],
        X[test_indices],
        y[train_indices],
        y[val_indices],
        y[test_indices]
    )


X_train, X_val, X_test, y_train, y_val, y_test = split_data(X, y)


# ==========================================
# 3. STANDARDIZATION
# ==========================================

def standardize(X_train, X_val, X_test): #z score

    mean = np.mean(X_train, axis=0)
    std = np.std(X_train, axis=0)

    std[std == 0] = 1

    X_train = (X_train - mean) / std
    X_val = (X_val - mean) / std
    X_test = (X_test - mean) / std

    return X_train, X_val, X_test


X_train, X_val, X_test = standardize(
    X_train, X_val, X_test
)


# ==========================================
# 4. RANDOM OVERSAMPLING
# ==========================================

def random_oversampling(X, y):

    class_0 = np.where(y == 0)[0]
    class_1 = np.where(y == 1)[0]

    if len(class_0) < len(class_1):
        minority = class_0
        majority = class_1
    else:
        minority = class_1
        majority = class_0

    extra = len(majority) - len(minority)

    extra_indices = np.random.choice(
        minority,
        size=extra,
        replace=True
    )

    new_indices = np.concatenate([
        majority,
        minority,
        extra_indices
    ])

    np.random.shuffle(new_indices)

    return X[new_indices], y[new_indices]


# ==========================================
# 5. LINEAR REGRESSION USING SGD
# ==========================================

class LinearRegressionSGD:

    def __init__(self, learning_rate=0.01, epochs=1000):

        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X, y):

        X = np.c_[np.ones(X.shape[0]), X]

        self.weights = np.zeros(X.shape[1])

        for epoch in range(self.epochs):

            indices = np.random.permutation(len(X))

            for i in indices:

                xi = X[i]
                yi = y[i]

                prediction = xi @ self.weights

                error = prediction - yi

                gradient = 2 * error * xi

                self.weights -= self.lr * gradient

    def predict(self, X):

        X = np.c_[np.ones(X.shape[0]), X]

        return X @ self.weights


# ==========================================
# 6. EVALUATION METRICS
# ==========================================

def calculate_metrics(y_actual, y_pred):

    TP = np.sum((y_actual == 1) & (y_pred == 1))
    TN = np.sum((y_actual == 0) & (y_pred == 0))
    FP = np.sum((y_actual == 0) & (y_pred == 1))
    FN = np.sum((y_actual == 1) & (y_pred == 0))

    accuracy = (TP + TN) / len(y_actual) #	Overall correct predictions

    precision = TP / (TP + FP + 1e-10) # How many predicted positive samples were actually positive

    recall = TP / (TP + FN + 1e-10) # How many actual positive samples were correctly identified

    f1 = 2 * precision * recall / (precision + recall + 1e-10) # 	Balance between precision and recall

    return accuracy, precision, recall, f1


# ==========================================
# 7. RUN THREE CLASSIFIERS
# ==========================================

def run_experiment(): # oversampling used for training

    categories = {
        0: "Iris-setosa",
        1: "Iris-versicolor",
        2: "Iris-virginica"
    }

    results = []

    for category in range(3):

        print("\n================================")
        print("Classifier:", categories[category])
        print("================================")

        # Convert into binary classification
        y_train_binary = (y_train == category).astype(int)
        y_val_binary = (y_val == category).astype(int)
        y_test_binary = (y_test == category).astype(int)

        # Oversample training data
        X_over, y_over = random_oversampling(
            X_train, y_train_binary
        )

        print("Before oversampling:", np.bincount(y_train_binary))
        print("After oversampling:", np.bincount(y_over))

        # Create model
        model = LinearRegressionSGD(
            learning_rate=0.01,
            epochs=1000
        )

        # Train model
        model.fit(X_over, y_over)

        # Predictions
        train_pred = (model.predict(X_train) >= 0.5).astype(int)

        val_pred = (model.predict(X_val) >= 0.5).astype(int)

        test_pred = (model.predict(X_test) >= 0.5).astype(int)

        # Calculate test metrics
        acc, prec, rec, f1 = calculate_metrics(
            y_test_binary, test_pred
        )

        print("Training Accuracy:",
              calculate_metrics(y_train_binary, train_pred)[0])

        print("Validation Accuracy:",
              calculate_metrics(y_val_binary, val_pred)[0])

        print("Testing Accuracy:", acc)
        print("Precision:", prec)
        print("Recall:", rec)
        print("F1 Score:", f1)

        results.append({
            "Category": categories[category],
            "Accuracy": acc,
            "Precision": prec,
            "Recall": rec,
            "F1 Score": f1
        })

    return pd.DataFrame(results)


# ==========================================
# 8. DISPLAY COMPARISON
# ==========================================

result = run_experiment()

print("\n========== FINAL COMPARISON ==========")

print(result.to_string(index=False))

result.to_csv("Question2_Oversampling.csv", index=False)
```

3. Use SMOTE technique to create synthetic examples. Create a linear regression model
on this oversampled dataset. Do the experiment for each of the categories and create
classifiers. Compare their performance. Use two settings for SMOTE.
a. Take any 2 samples from the minority class and generate the new synthetic
points.
b. Find the nearest sample for any sample and then apply the SMOTE technique.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# ==========================================
# 1. LOAD DATASET
# ==========================================

df = pd.read_csv("Iris.csv")

X = df.drop(["Id", "Species"], axis=1).values

y = df["Species"].map({
    "Iris-setosa": 0,
    "Iris-versicolor": 1,
    "Iris-virginica": 2
}).values


# ==========================================
# 2. SPLIT DATASET
# ==========================================

def split_data(X, y):

    train_indices = []
    val_indices = []
    test_indices = []

    for category in np.unique(y):

        indices = np.where(y == category)[0]

        np.random.shuffle(indices)

        n = len(indices)

        train_size = int(0.80 * n)
        val_size = int(0.10 * n)

        train_indices.extend(indices[:train_size])
        val_indices.extend(indices[train_size:train_size + val_size])
        test_indices.extend(indices[train_size + val_size:])

    np.random.shuffle(train_indices)
    np.random.shuffle(val_indices)
    np.random.shuffle(test_indices)

    return (
        X[train_indices],
        X[val_indices],
        X[test_indices],
        y[train_indices],
        y[val_indices],
        y[test_indices]
    )


X_train, X_val, X_test, y_train, y_val, y_test = split_data(X, y)


# ==========================================
# 3. STANDARDIZATION
# ==========================================

def standardize(X_train, X_val, X_test):

    mean = np.mean(X_train, axis=0)
    std = np.std(X_train, axis=0)

    std[std == 0] = 1

    X_train = (X_train - mean) / std
    X_val = (X_val - mean) / std
    X_test = (X_test - mean) / std

    return X_train, X_val, X_test


X_train, X_val, X_test = standardize(
    X_train, X_val, X_test
)


# ==========================================
# 4. SMOTE SETTING A
# RANDOM PAIR OF MINORITY SAMPLES
# ==========================================

def smote_random_pair(X, y):

    class_0 = np.where(y == 0)[0]
    class_1 = np.where(y == 1)[0]

    if len(class_0) < len(class_1):
        minority = class_0
        majority = class_1
    else:
        minority = class_1
        majority = class_0

    extra = len(majority) - len(minority)

    synthetic_samples = []

    for i in range(extra):

        # Select two different minority samples
        idx = np.random.choice(
            minority,
            size=2,
            replace=False
        )

        sample1 = X[idx[0]]
        sample2 = X[idx[1]]

        # Random interpolation
        lam = np.random.random()

        new_sample = sample1 + lam * (sample2 - sample1)

        synthetic_samples.append(new_sample)

    synthetic_samples = np.array(synthetic_samples)

    X_new = np.vstack([X, synthetic_samples])

    y_new = np.concatenate([
        y,
        np.full(extra, y[minority[0]])
    ])

    return X_new, y_new


# ==========================================
# 5. SMOTE SETTING B
# NEAREST MINORITY NEIGHBOR
# ==========================================

def smote_nearest_neighbor(X, y):

    class_0 = np.where(y == 0)[0]
    class_1 = np.where(y == 1)[0]

    if len(class_0) < len(class_1):
        minority = class_0
        majority = class_1
    else:
        minority = class_1
        majority = class_0

    extra = len(majority) - len(minority)

    synthetic_samples = []

    for i in range(extra):

        # Select a random minority sample
        idx = np.random.choice(minority)

        sample1 = X[idx]

        # Calculate distances
        distances = np.linalg.norm(
            X[minority] - sample1,
            axis=1
        )

        # Ignore itself
        distances[minority == idx] = np.inf

        # Find nearest neighbor
        nearest_position = np.argmin(distances)

        nearest_idx = minority[nearest_position]

        sample2 = X[nearest_idx]

        # Generate synthetic sample
        lam = np.random.random()

        new_sample = sample1 + lam * (sample2 - sample1)

        synthetic_samples.append(new_sample)

    synthetic_samples = np.array(synthetic_samples)

    X_new = np.vstack([X, synthetic_samples])

    y_new = np.concatenate([
        y,
        np.full(extra, y[minority[0]])
    ])

    return X_new, y_new


# ==========================================
# 6. LINEAR REGRESSION USING SGD
# ==========================================

class LinearRegressionSGD:

    def __init__(self, learning_rate=0.01, epochs=1000):

        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X, y):

        X = np.c_[np.ones(X.shape[0]), X]

        self.weights = np.zeros(X.shape[1])

        for epoch in range(self.epochs):

            indices = np.random.permutation(len(X))

            for i in indices:

                xi = X[i]
                yi = y[i]

                prediction = xi @ self.weights

                error = prediction - yi

                gradient = 2 * error * xi

                self.weights -= self.lr * gradient

    def predict(self, X):

        X = np.c_[np.ones(X.shape[0]), X]

        return X @ self.weights


# ==========================================
# 7. EVALUATION METRICS
# ==========================================

def calculate_metrics(y_actual, y_pred):

    TP = np.sum((y_actual == 1) & (y_pred == 1))
    TN = np.sum((y_actual == 0) & (y_pred == 0))
    FP = np.sum((y_actual == 0) & (y_pred == 1))
    FN = np.sum((y_actual == 1) & (y_pred == 0))

    accuracy = (TP + TN) / len(y_actual)

    precision = TP / (TP + FP + 1e-10)

    recall = TP / (TP + FN + 1e-10)

    f1 = 2 * precision * recall / (precision + recall + 1e-10)

    return accuracy, precision, recall, f1


# ==========================================
# 8. RUN SMOTE EXPERIMENT
# ==========================================

def run_smote_experiment(smote_function, method_name):

    categories = {
        0: "Iris-setosa",
        1: "Iris-versicolor",
        2: "Iris-virginica"
    }

    results = []

    for category in range(3):

        print("\n================================")
        print(method_name)
        print("Classifier:", categories[category])
        print("================================")

        # Binary labels
        y_train_binary = (y_train == category).astype(int)
        y_val_binary = (y_val == category).astype(int)
        y_test_binary = (y_test == category).astype(int)

        # Apply SMOTE to training data only
        X_smote, y_smote = smote_function(
            X_train, y_train_binary
        )

        print("Before SMOTE:", np.bincount(y_train_binary))
        print("After SMOTE:", np.bincount(y_smote))

        # Create model
        model = LinearRegressionSGD(
            learning_rate=0.01,
            epochs=1000
        )

        # Train
        model.fit(X_smote, y_smote)

        # Predictions
        train_pred = (model.predict(X_train) >= 0.5).astype(int)

        val_pred = (model.predict(X_val) >= 0.5).astype(int)

        test_pred = (model.predict(X_test) >= 0.5).astype(int)

        # Evaluation
        acc, prec, rec, f1 = calculate_metrics(
            y_test_binary, test_pred
        )

        print("Training Accuracy:",
              calculate_metrics(y_train_binary, train_pred)[0])

        print("Validation Accuracy:",
              calculate_metrics(y_val_binary, val_pred)[0])

        print("Testing Accuracy:", acc)
        print("Precision:", prec)
        print("Recall:", rec)
        print("F1 Score:", f1)

        results.append({
            "Method": method_name,
            "Category": categories[category],
            "Accuracy": acc,
            "Precision": prec,
            "Recall": rec,
            "F1 Score": f1
        })

    return pd.DataFrame(results)


# ==========================================
# 9. RUN SETTING A
# ==========================================

result_A = run_smote_experiment(
    smote_random_pair,
    "SMOTE Random Pair"
)


# ==========================================
# 10. RUN SETTING B
# ==========================================

result_B = run_smote_experiment(
    smote_nearest_neighbor,
    "SMOTE Nearest Neighbor"
)


# ==========================================
# 11. FINAL COMPARISON
# ==========================================

final_result = pd.concat(
    [result_A, result_B],
    ignore_index=True
)

print("\n========== SMOTE COMPARISON ==========")

print(final_result.to_string(index=False))

final_result.to_csv("Question3_SMOTE_Comparison.csv", index=False)
```

4. Design a logistic regression model with stochastic gradient descent on this dataset.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# ==========================================
# 1. LOAD DATASET
# ==========================================

df = pd.read_csv("Iris.csv")

X = df.drop(["Id", "Species"], axis=1).values

y = df["Species"].map({
    "Iris-setosa": 0,
    "Iris-versicolor": 1,
    "Iris-virginica": 2
}).values


# ==========================================
# 2. SPLIT DATASET
# ==========================================

def split_data(X, y):

    train_indices = []
    val_indices = []
    test_indices = []

    for category in np.unique(y):

        indices = np.where(y == category)[0]

        np.random.shuffle(indices)

        n = len(indices)

        train_size = int(0.80 * n)
        val_size = int(0.10 * n)

        train_indices.extend(indices[:train_size])
        val_indices.extend(indices[train_size:train_size + val_size])
        test_indices.extend(indices[train_size + val_size:])

    np.random.shuffle(train_indices)
    np.random.shuffle(val_indices)
    np.random.shuffle(test_indices)

    return (
        X[train_indices],
        X[val_indices],
        X[test_indices],
        y[train_indices],
        y[val_indices],
        y[test_indices]
    )


X_train, X_val, X_test, y_train, y_val, y_test = split_data(X, y)


# ==========================================
# 3. STANDARDIZATION
# ==========================================

def standardize(X_train, X_val, X_test):

    mean = np.mean(X_train, axis=0)
    std = np.std(X_train, axis=0)

    std[std == 0] = 1

    X_train = (X_train - mean) / std
    X_val = (X_val - mean) / std
    X_test = (X_test - mean) / std

    return X_train, X_val, X_test


X_train, X_val, X_test = standardize(
    X_train, X_val, X_test
)


# ==========================================
# 4. SIGMOID FUNCTION
# ==========================================

def sigmoid(z):

    z = np.clip(z, -500, 500)

    return 1 / (1 + np.exp(-z))


# ==========================================
# 5. LOGISTIC REGRESSION USING SGD
# ==========================================

class LogisticRegressionSGD:

    def __init__(self, learning_rate=0.01, epochs=1000):

        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X, y):

        # Add intercept
        X = np.c_[np.ones(X.shape[0]), X]

        # Initialize weights
        self.weights = np.zeros(X.shape[1])

        # SGD training
        for epoch in range(self.epochs):

            indices = np.random.permutation(len(X))

            for i in indices:

                xi = X[i]
                yi = y[i]

                # Linear calculation
                z = xi @ self.weights

                # Sigmoid probability
                prediction = sigmoid(z)

                # Gradient
                gradient = (prediction - yi) * xi

                # Update weights
                self.weights -= self.lr * gradient

    def predict_proba(self, X):

        X = np.c_[np.ones(X.shape[0]), X]

        z = X @ self.weights

        return sigmoid(z)

    def predict(self, X):

        probabilities = self.predict_proba(X)

        return (probabilities >= 0.5).astype(int)


# ==========================================
# 6. EVALUATION METRICS
# ==========================================

def calculate_metrics(y_actual, y_pred):

    TP = np.sum((y_actual == 1) & (y_pred == 1))
    TN = np.sum((y_actual == 0) & (y_pred == 0))
    FP = np.sum((y_actual == 0) & (y_pred == 1))
    FN = np.sum((y_actual == 1) & (y_pred == 0))

    accuracy = (TP + TN) / len(y_actual)

    precision = TP / (TP + FP + 1e-10)

    recall = TP / (TP + FN + 1e-10)

    f1 = 2 * precision * recall / (precision + recall + 1e-10)

    return accuracy, precision, recall, f1


# ==========================================
# 7. CREATE THREE CLASSIFIERS
# ==========================================

def run_logistic_experiment():

    categories = {
        0: "Iris-setosa",
        1: "Iris-versicolor",
        2: "Iris-virginica"
    }

    results = []

    for category in range(3):

        print("\n================================")
        print("Classifier:", categories[category])
        print("================================")

        # Binary labels
        y_train_binary = (y_train == category).astype(int)
        y_val_binary = (y_val == category).astype(int)
        y_test_binary = (y_test == category).astype(int)

        # Create model
        model = LogisticRegressionSGD(
            learning_rate=0.01,
            epochs=1000
        )

        # Train model
        model.fit(X_train, y_train_binary)

        # Predictions
        train_pred = model.predict(X_train)

        val_pred = model.predict(X_val)

        test_pred = model.predict(X_test)

        # Evaluation
        acc, prec, rec, f1 = calculate_metrics(
            y_test_binary, test_pred
        )

        print("Training Accuracy:",
              calculate_metrics(y_train_binary, train_pred)[0])

        print("Validation Accuracy:",
              calculate_metrics(y_val_binary, val_pred)[0])

        print("Testing Accuracy:", acc)
        print("Precision:", prec)
        print("Recall:", rec)
        print("F1 Score:", f1)

        results.append({
            "Category": categories[category],
            "Accuracy": acc,
            "Precision": prec,
            "Recall": rec,
            "F1 Score": f1
        })

    return pd.DataFrame(results)


# ==========================================
# 8. DISPLAY FINAL RESULTS
# ==========================================

result = run_logistic_experiment()

print("\n========== LOGISTIC REGRESSION RESULTS ==========")

print(result.to_string(index=False))

result.to_csv("Question4_Logistic_Regression.csv", index=False)
```
